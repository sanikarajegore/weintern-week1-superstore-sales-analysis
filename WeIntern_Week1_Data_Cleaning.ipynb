```python

total_missing = df.isnull().sum().sum()

print("Total Missing Values:", total_missing)

total_missing = df.isnull().sum().sum()

print("Total Missing Values:", total_missing)
```

```python

duplicate_count = df.duplicated().sum()

print("Total Duplicate Rows:", duplicate_count)
```

```python

duplicates = df[df.duplicated()]

print("Duplicate Records:")
display(duplicates.head())

```

```python

# Remove duplicate rows

df = df.drop_duplicates()

print("Duplicate records removed successfully!")
print("Updated dataset shape:", df.shape)
```

```python

missing_values = df.isnull().sum()

print(missing_values[missing_values > 0])
```

```python

# Fill missing categorical values
for col in df.select_dtypes(include=['object']).columns:
    df[col] = df[col].fillna("Unknown")

# Fill missing numeric values with median
for col in df.select_dtypes(include=np.number).columns:
    df[col] = df[col].fillna(df[col].median())

print("Missing values handled!")
print("Remaining missing values:", df.isnull().sum().sum())
```

```python

for col in df.select_dtypes(include=['object']).columns:
    df[col] = df[col].str.strip()

print("Text columns standardized successfully!")
```

```python

print(df.dtypes)
```

```python

df['Order Date'] = pd.to_datetime(
    df['Order Date'], errors='coerce'
)

df['Ship Date'] = pd.to_datetime(
    df['Ship Date'], errors='coerce'
)

print("Date columns converted successfully!")
print(df[['Order Date', 'Ship Date']].dtypes)
```

```python

print("FINAL DATA QUALITY REPORT")

print("Dataset Shape:", df.shape)

print("Missing Values:", df.isnull().sum().sum())

print("Duplicate Rows:", df.duplicated().sum())

print("\nFinal Dataset Preview:")
display(df.head())
```

```python

from google.colab import files
import pandas as pd
import numpy as np

uploaded = files.upload()

df = pd.read_csv('Cleaned_Sample_Superstore.csv')

print("Cleaned dataset loaded successfully!")
print("Dataset Shape:", df.shape)
```

```python

total_revenue = df['Sales'].sum()

print("Total Revenue:", round(total_revenue, 2))
```

```python

total_orders = df['Order ID'].nunique()

print("Total Orders:", total_orders)
```

```python

average_order_value = total_revenue / total_orders

print("Average Order Value:", round(average_order_value, 2))
```

```python
total_units_sold = df['Order Quantity'].sum()

print("Total Units Sold:", total_units_sold)
```

```python
category_sales = df.groupby('Product Category')['Order Quantity'].sum()

print("Units Sold by Category:")
display(category_sales.sort_values(ascending=False))

print("Top Selling Category:", category_sales.idxmax())
```

```python

product_revenue = df.groupby('Product Name')['Sales'].sum()

print("Top 10 Products by Revenue:")

display(
    product_revenue
    .sort_values(ascending=False)
    .head(10)
)
```

```python

df['Order Date'] = pd.to_datetime(df['Order Date'])

df['Month'] = df['Order Date'].dt.month_name()

monthly_sales = df.groupby('Month')['Sales'].sum()

print("Monthly Sales:")
display(monthly_sales)

print("Highest Sales Month:",
      monthly_sales.idxmax())
```

```python

total_profit = df['Profit'].sum()

print("Total Profit:", round(total_profit, 2))
```

```python

kpi_summary = pd.DataFrame({
    'Metric': [
        'Total Revenue',
        'Total Orders',
        'Average Order Value',
        'Total Units Sold',
        'Total Profit',
        'Top Selling Category',
        'Highest Sales Month'
    ],
    'Value': [
        round(total_revenue, 2),
        total_orders,
        round(average_order_value, 2),
        total_units_sold,
        round(total_profit, 2),
        category_sales.idxmax(),
        monthly_sales.idxmax()
    ]
})

display(kpi_summary)
```

```python

category_analysis = df.groupby(
    'Product Category'
)['Order Quantity'].sum().sort_values(
    ascending=False
)

print("Units Sold by Product Category:")
display(category_analysis)

print("Best Selling Category:",
      category_analysis.idxmax())
```

```python

top_products = df.groupby(
    'Product Name'
)['Sales'].sum().sort_values(
    ascending=False
).head(10)

print("Top 10 Revenue Generating Products:")
display(top_products)
```

```python

df['Order Date'] = pd.to_datetime(df['Order Date'])

df['Month'] = df['Order Date'].dt.to_period('M')

monthly_sales = df.groupby('Month')['Sales'].sum()

print("Monthly Sales Analysis:")
display(monthly_sales)

print("Best Sales Month:",
      monthly_sales.idxmax())
```

```python

category_profit = df.groupby(
    'Product Category'
)['Profit'].sum().sort_values(
    ascending=False
)

print("Profit by Category:")
display(category_profit)

print("Most Profitable Category:",
      category_profit.idxmax())
```

```python

print("BUSINESS INSIGHTS")

print("1. Total revenue:", round(total_revenue, 2))

print("2. Total profit:", round(total_profit, 2))

print("3. Average order value:",
      round(average_order_value, 2))

print("4. Highest selling category:",
      category_analysis.idxmax())

print("5. Highest sales month:",
      monthly_sales.idxmax())

print("6. Most profitable category:",
      category_profit.idxmax())

print("7. Total units sold:", total_units_sold)

print("8. Total unique orders:", total_orders)
```

```python

kpi_summary.to_csv('Sales_KPI_Summary.csv', index=False)

category_analysis.to_csv('Category_Analysis.csv')

top_products.to_csv('Top_10_Products.csv')

monthly_sales.to_csv('Monthly_Sales.csv')

category_profit.to_csv('Category_Profit.csv')

print("All analysis files saved successfully!")
```

```python

category_revenue = df.groupby(
    'Product Category'
)['Sales'].sum().sort_values(ascending=False)

plt.figure(figsize=(9, 5))

category_revenue.plot(
    kind='bar',
    title='Total Sales by Product Category'
)

plt.xlabel('Product Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig('Sales_by_Category.png', dpi=300)
plt.show()
```

```python

plt.figure(figsize=(12, 5))

monthly_sales.plot(
    kind='line',
    marker='o',
    title='Monthly Sales Trend'
)

plt.xlabel('Month')
plt.ylabel('Sales Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('Monthly_Sales_Trend.png', dpi=300)
plt.show()
```

```python

plt.figure(figsize=(8, 8))

category_analysis.plot(
    kind='pie',
    autopct='%1.1f%%',
    startangle=90,
    title='Units Sold by Product Category'
)

plt.ylabel('')
plt.tight_layout()
plt.savefig('Category_Distribution.png', dpi=300)
plt.show()
```

```python

fig, axes = plt.subplots(2, 2, figsize=(16, 10))

# Chart 1: Revenue by category
category_revenue.plot(
    kind='bar', ax=axes[0, 0],
    title='Revenue by Category'
)
axes[0, 0].set_xlabel('Category')
axes[0, 0].set_ylabel('Sales')

# Chart 2: Monthly sales
monthly_sales.plot(
    kind='line', marker='o',
    ax=axes[0, 1],
    title='Monthly Sales Trend'
)
axes[0, 1].set_ylabel('Sales')

# Chart 3: Units sold
category_analysis.plot(
    kind='pie',
    autopct='%1.1f%%',
    ax=axes[1, 0],
    title='Units Sold by Category'
)
axes[1, 0].set_ylabel('')

# Chart 4: Profit by category
category_profit.plot(
    kind='bar',
    ax=axes[1, 1],
    title='Profit by Category'
)
axes[1, 1].set_ylabel('Profit')
axes[1, 1].tick_params(axis='x', rotation=20)

plt.suptitle(
    'SAMPLE SUPERSTORE SALES DASHBOARD',
    fontsize=17,
    fontweight='bold'
)

plt.tight_layout()
plt.savefig('Superstore_Dashboard.png', dpi=300)
plt.show()
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

category_revenue = df.groupby(
    'Product Category'
)['Sales'].sum().sort_values(ascending=False)

plt.figure(figsize=(9, 5))

category_revenue.plot(kind='bar')

plt.title('Total Sales by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)

plt.tight_layout()
plt.savefig('Sales_by_Category.png', dpi=300)
plt.show()
```
